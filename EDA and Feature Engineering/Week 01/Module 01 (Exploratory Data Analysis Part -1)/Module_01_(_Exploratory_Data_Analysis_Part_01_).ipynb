# Univariate Analysis (Categorical Value Analysis -1)

### Survived

```python
sns.countplot(data=df, x='Survived')

# plt.show()
```

```python
df['Survived'].value_counts() / len(df) * 100     # % - percentage ber kora
```

### Pie chart


```python
survived_count = df['Survived'].value_counts()

survived_label = df['Survived'].unique()

plt.pie(survived_count,labels=survived_label, autopct="%1.1f")

plt.title('Pie Chart of survived rate')
plt.show()
```

# Univariate Analysis (Numerical Column Analysis - 1)

### Age

```python
sns.histplot(df['Age'], bins=50)

plt.title('Histogram of Age')

plt.xlabel('Age')
plt.ylabel('frequency')
plt.show()
```

```python
# replot -- kde plot

sns.kdeplot(df['Age'])

plt.title('Histogram of Age')

plt.xlabel('Age')
plt.ylabel('Probability distribution')
plt.show()
```
